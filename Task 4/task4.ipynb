```python
import requests
import pandas as pd
```

```python

# Step 2: Your TMDB API Key
API_KEY = 'api_key'
BASE_URL = 'https://api.themoviedb.org/3/movie/popular'
```

```python
def fetch_movie_data():
    # Step 3: Fetch popular movies data using requests
    params = {
        'api_key': API_KEY,
        'language': 'en-US',
        'page': 1
    }
    
    response = requests.get(BASE_URL, params=params)
    
    if response.status_code == 200:
        data = response.json()
        movies_list = data.get('results', [])
        
        # Step 3 (cont.): Extract fields
        extracted_data = []
        for movie in movies_list:
            movie_info = {
                'Movie title': movie.get('title'),
                'Release date': movie.get('release_date'),
                'Rating': movie.get('vote_average'),
                'Popularity': movie.get('popularity'),
                # Extra column of choice:
                'Overview': movie.get('overview') 
            }
            extracted_data.append(movie_info)
        
        # Step 4: Convert the API response into a Pandas DataFrame
        df = pd.DataFrame(extracted_data)
        
        # Step 5: Save the collected data into a CSV file
        df.to_csv('tmdb_movies.csv', index=False)
        print("Success! Data saved to tmdb_movies.csv")
        return df
    else:
        print(f"Error: {response.status_code}")
        return None
```

```python
fetch_movie_data()
```
