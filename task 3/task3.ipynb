```python
import sqlite3
import pandas as pd

```

```python
# 1. Create a SQLite database (sample.db)
# Connecting to the file will automatically create it if it doesn't exist
conn = sqlite3.connect("sample.db")
cursor = conn.cursor()
```

```python

# 2. Create a table named employees with the specified columns
cursor.execute(
    """
    CREATE TABLE IF NOT EXISTS employees (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        department TEXT NOT NULL,
        salary REAL NOT NULL
    )
"""
)

```

```python
# 3. Insert at least 5 records
# We use a list of tuples and executemany for clean insertion
employee_data = [
    (1, "Akshit", "Engineering", 85000.00),
    (2, "Preeti", "Marketing", 62000.00),
    (3, "Chirag", "Sales", 55000.00),
    (4, "Divya", "Engineering", 95000.00),
    (5, "Vikas", "HR", 60000.00),
    (6, "Akshar", "Executive", 50000.00),
    (7, "Rahul", "Warehouse", 65000.00),
]

```

```python
# 'INSERT OR IGNORE' prevents errors if you run this script multiple times
cursor.executemany(
    "INSERT OR IGNORE INTO employees (id, name, department, salary) VALUES (?, ?, ?, ?)",
    employee_data,
)
conn.commit()

```

```python
# 4. Read the table into Pandas using a SQL query
query = "SELECT * FROM employees"
df = pd.read_sql_query(query, conn)

# 5. Display the DataFrame
print("            Employees DataFrame")
print("-------------------------------------------")
print(df)

conn.close()
```
